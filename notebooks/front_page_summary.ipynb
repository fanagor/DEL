# Front-Page Summary

Combine the selected arithmetic results and results for the universal learning task in the paper's summary plot. Edit `front_page` in `paper/figures.yaml` to change the selected runs. All five selected records require final evaluations.

Run all cells to export the figure as PDF, SVG and PNG using the shared paper style. Outputs go to `analysis_outputs/paper_figures/front_page_summary/`. `DEL_ANALYSIS_DIR` selects another figure root and preserves this subfolder. This notebook reads saved evaluations and does not train or evaluate models.


```python
from pathlib import Path
import os, sys
import matplotlib.pyplot as plt
from IPython.display import display
ROOT = Path(os.environ["DEL_REPO_ROOT"]) if os.environ.get("DEL_REPO_ROOT") else next(
    p for p in (Path.cwd(), *Path.cwd().parents) if (p / "src/differentiable_exact_learning").is_dir())
sys.path.insert(0, str(ROOT / "src"))
from differentiable_exact_learning.analysis.release import final_results_table
from differentiable_exact_learning.analysis.paper_heatmaps import save_figure, figure_output_dir
from differentiable_exact_learning.analysis.plot_style import apply_paper_style
apply_paper_style()
OUTPUT_DIR = figure_output_dir(ROOT, 'front_page_summary')
def export(fig, name):
    paths = save_figure(fig, OUTPUT_DIR / name)
    display(fig)
    plt.close(fig)
    return paths
```

```python
import yaml
from differentiable_exact_learning.analysis.release import load_release_record
from differentiable_exact_learning.analysis.front_page import plot_front_page
selection = yaml.safe_load((ROOT / "paper/figures.yaml").read_text())["front_page"]
front = {task: load_release_record(ROOT, path) for task, path in selection.items()}
display(final_results_table(list(front.values())))
export(plot_front_page({task: front[task] for task in ("add2", "add10", "mult2", "mult10")},
                       front["universal"]), "front_page")
final_results_table(list(front.values())).to_csv(OUTPUT_DIR / "front_page_summary.csv", index=False)
```
